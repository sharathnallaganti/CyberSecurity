# Nmap for Cybersecurity

Nmap stands for **Network Mapper**. It is an open-source tool used to discover devices, identify accessible network services, and collect information for security assessments.

Security professionals use Nmap to understand what a system exposes to the network.




## 1. What Can Nmap Do?

Nmap helps answer these questions:

1. Which devices appear to be online?
2. Which ports are accessible?
3. Which services are listening on those ports?
4. What software and versions might those services use?
5. What operating system might a device run?
6. Does filtering prevent access to certain services?

### Example

A server is supposed to provide a website, but a scan also reveals an accessible database.

A security analyst would investigate:

- Is the database supposed to be accessible?
- Who needs access?
- Should firewall rules restrict it?
- Is the service securely configured and maintained?

**An open port is not automatically a vulnerability.** It is an observation that may require investigation.



## 2. Essential Networking Concepts

### 2.1 Host

A host is a device or system connected to a network.

Examples include:

- Laptops
- Servers
- Routers
- Virtual machines

### 2.2 IP — Internet Protocol Address

An IP address identifies a network interface for communication.

Example:

```text
192.168.1.10
```

The special address:

```text
127.0.0.1
```

is a loopback address. It refers to the computer or runtime executing the command.

### 2.3 Port

A port is a numbered communication endpoint.

A single server can offer multiple services using different ports.

### 2.4 Service

A service is software listening for network communication.

Examples include:

- A web server
- A remote administration service
- A database server

### 2.5 Firewall

A firewall applies rules that allow or block network traffic.

### Building Analogy

| Networking concept | Building analogy |
|---|---|
| IP address | Building address |
| Port | Numbered entrance |
| Service | Department behind the entrance |
| Firewall | Rules controlling entry |
| Nmap | A tool checking which entrances respond |



## 3. TCP and UDP

### 3.1 TCP — Transmission Control Protocol

TCP establishes connections and provides reliable, ordered delivery of data.

It is commonly used for web traffic, remote administration, and file transfers.

### 3.2 UDP — User Datagram Protocol

UDP sends individual messages without TCP's connection setup or built-in delivery guarantees.

It is commonly used for name resolution, real-time communication, and other services.

### Important Difference

TCP and UDP have separate port spaces.

```text
53/tcp
53/udp
```

These are two distinct endpoints, even though they share the same port number.



## 4. Common Ports and Services

| Port | Protocol | Common service | Purpose |
|---|---|---|---|
| 22 | TCP | SSH — Secure Shell | Secure remote administration |
| 80 | TCP | HTTP — Hypertext Transfer Protocol | Web communication |
| 443 | TCP | HTTPS — Hypertext Transfer Protocol Secure | Encrypted web communication |
| 53 | TCP and UDP | DNS — Domain Name System | Name resolution |
| 445 | TCP | SMB — Server Message Block | File sharing |
| 3389 | TCP | RDP — Remote Desktop Protocol | Remote desktop access |
| 123 | UDP | NTP — Network Time Protocol | Time synchronization |
| 161 | UDP | SNMP — Simple Network Management Protocol | Network monitoring and management |

> These are common associations. Services can run on nonstandard ports.
>
> A port number alone does not prove which application is running.


## 6. Basic Command Structure

```bash
nmap [options] target
```

Example:

```bash
nmap -sV -p 80,443 127.0.0.1
```

### Explanation

| Component | Meaning |
|---|---|
| `nmap` | Start Nmap |
| `-sV` | Attempt service and version detection |
| `-p 80,443` | Scan ports 80 and 443 |
| `127.0.0.1` | Scan the local runtime |

Options are case-sensitive.

For example:

- `-sV` means service/version detection.
- `-sU` means UDP scanning.

## 7. Your First Scan

```bash
nmap 127.0.0.1
```

It does not automatically scan:

- Every TCP port
- UDP ports
- Every possible security vulnerability

### Illustrative Output

```text
PORT      STATE     SERVICE
22/tcp    open      ssh
80/tcp    closed    http
443/tcp   filtered  https
```

This output is an example, not a result from your computer.

---

## 8. Understanding Port States

| State | Meaning |
|---|---|
| `open` | A service accepts communication on the port |
| `closed` | The port is reachable, but no service is listening |
| `filtered` | Filtering prevents a definite open/closed decision |
| `unfiltered` | The probe reaches the port, but the scan cannot determine whether it is open |
| `open\|filtered` | Nmap cannot distinguish open from filtered |
| `closed\|filtered` | Nmap cannot distinguish closed from filtered |

### Open

```text
22/tcp open ssh
```

A service is listening on TCP port 22.

The service label may be based on the port's usual association unless service detection was performed.

### Closed

```text
80/tcp closed http
```

The port responds, but no service is listening.

### Filtered

```text
443/tcp filtered https
```

Nmap cannot determine whether the service is listening because filtering interferes with the probes or responses.

### Important Point

Port states depend on where you scan from.

A service may appear:

- Open from an internal network
- Filtered from outside the network

## 9. Selecting Ports

### 9.1 Scan One Port

```bash
nmap -p 80 127.0.0.1
```

### 9.2 Scan Several Ports

```bash
nmap -p 22,80,443 127.0.0.1
```

### 9.3 Scan a Numeric Range

```bash
nmap -p 1-1000 127.0.0.1
```

This scans ports numbered 1 through 1000.

### 9.4 Scan the Most Common 100 Ports

```bash
nmap -F 127.0.0.1
```

### 9.5 Scan TCP Ports 1–65535

```bash
nmap -p- 127.0.0.1
```

Port zero is excluded unless explicitly requested.

### Comparison

| Option | Selection |
|---|---|
| Default | 1,000 most common TCP ports |
| `-F` | Normally 100 most common TCP ports |
| `-p 1-1000` | Ports numbered 1–1000 |
| `-p 80,443` | Only ports 80 and 443 |
| `-p-` | Ports 1–65535 for the selected protocol |

## 10. Host Discovery

Host discovery asks:

> Which devices appear to be online?

### Example for an Authorized Local Lab

```bash
nmap -sn 192.168.1.0/24
```

`-sn` performs host discovery without a port scan.

Do not use this example to scan Kaggle's network infrastructure.

### What Does /24 Mean?

CIDR stands for **Classless Inter-Domain Routing**.

The range:

```text
192.168.1.0/24
```

describes 256 addresses, from:

```text
192.168.1.0
```

through:

```text
192.168.1.255
```

### How Discovery Works

Depending on the environment, Nmap may use:

- ARP — Address Resolution Protocol
- ICMP — Internet Control Message Protocol
- TCP probes

Host discovery is not always just an ordinary ping.

---

## 11. Skipping Host Discovery

A known device might not answer discovery probes.

For an authorized lab host:

```bash
nmap -Pn -p 80,443 192.168.1.10
```

`-Pn` tells Nmap to attempt scanning without relying on the usual host-discovery decision.

### What -Pn Does Not Do

It does not:

- Bypass firewall rules
- Prove the device is online
- Guarantee that services are reachable

## 12. TCP Connect Scan

### Command

```bash
nmap -sT -p 80,443 127.0.0.1
```

### Meaning

`-sT` uses the operating system's connection mechanism to attempt TCP connections.

### TCP Three-Way Handshake

1. Client sends SYN — Synchronize.
2. Server replies with SYN/ACK — Synchronize/Acknowledgment.
3. Client sends ACK — Acknowledgment.

### Why Learn This First?

TCP connect scanning is useful when raw-packet privileges are unavailable.

It is a suitable starting point for a managed notebook environment.

---

## 13. TCP SYN Scan

### Command for Your Own Local Lab

```bash
sudo nmap -sS -p 80,443 127.0.0.1
```

### Meaning

`-sS` examines TCP connection setup without completing the normal handshake.

### Typical Responses

| Response | Typical interpretation |
|---|---|
| SYN/ACK | Open |
| RST — Reset | Closed |
| No response after retries or certain network errors | Filtered |

### Privileges

This scan typically requires elevated privileges.

`sudo` runs a command with elevated permissions.

A managed notebook may restrict raw packets even when its process runs with administrative privileges.

> SYN scanning is detectable. “Half-open” does not mean invisible.

---

## 14. UDP Scan

### Command for Your Own Local Lab

```bash
sudo nmap -sU -p 53,123,161 127.0.0.1
```

### Meaning

`-sU` scans UDP ports.

### Why UDP Can Be Difficult

UDP does not have TCP's connection handshake.

A service may ignore an unexpected probe even when it is listening.

### Possible Results

- A UDP response can indicate an open port.
- An ICMP port-unreachable response usually indicates a closed port.
- Silence may produce `open|filtered`.

UDP scans can be slower because Nmap may need to wait and retry.

---

## 15. Service and Version Detection

### Command

```bash
nmap -sV -p 80,443 127.0.0.1
```

### Purpose

Port scanning asks:

> Is something listening?

Service detection asks:

> What appears to be listening?

### Illustrative Output

```text
PORT    STATE SERVICE VERSION
80/tcp  open  http    nginx ...
```

Nmap sends additional probes and compares the responses against service fingerprints.

### Important Limitation

A detected version does not prove a vulnerability exists.

Before reporting a security issue, verify:

- Actual software
- Patch status
- Configuration
- Exposure
- Relevant security conditions

---

## 16. Operating-System Detection

OS stands for **Operating System**.

### Command

```bash
sudo nmap -O 127.0.0.1
```

`-O` uses the capital letter O.

### How It Works

Nmap compares characteristics of network responses with known operating-system fingerprints.

### Limitations

- Results are estimates.
- Firewalls and network behavior can affect accuracy.
- Detection works better when at least one open and one closed TCP port are available.

---

## 17. The -A Option

### Command

```bash
sudo nmap -A 127.0.0.1
```

This enables:

- Operating-system detection
- Service/version detection
- Default script scanning
- Traceroute

Traceroute attempts to identify the network path to the target.

### Important Distinction

`-A` does not mean:

> Scan all ports.

Port coverage still depends on your port-selection options.

Use focused commands while learning so you understand each operation.

## 18. NSE — Nmap Scripting Engine

NSE allows Nmap to run scripts that collect information or perform particular checks.

Scripts differ in behavior. Review the documentation before running them.

### Retrieve a Web Page Title

### Default Scripts

```bash
nmap -sC 127.0.0.1
```

`-sC` selects the default script set.

Default does not guarantee that every selected script is nonintrusive.

### Step 1: Import Modules and Choose an Available Port

The operating system chooses a free port. Your port number may differ each time.

```python
import socket
import subprocess
import threading
import http.server
import urllib.request

# Close a previous instance if you rerun this setup cell.
if "server" in globals() and server is not None:
    server.shutdown()
    server.server_close()
    server_thread.join(timeout=3)

if "reservation" in globals() and reservation is not None:
    reservation.close()

server = None

# Reserve a local port without listening for connections.
reservation = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
reservation.bind(("127.0.0.1", 0))
lab_port = reservation.getsockname()[1]

print("Practice port:", lab_port)
```

### Step 2: Create a Local Scan Function

```python
def scan_local(*extra_options):
    command = [
        "nmap",
        "-sT",
        "-Pn",
        "-n",
        "-p", str(lab_port),
        *extra_options,
        "127.0.0.1"
    ]

    result = subprocess.run(
        command,
        capture_output=True,
        text=True,
        timeout=120
    )

    print("Command:", " ".join(command))
    print(result.stdout)

    if result.stderr:
        print(result.stderr)

    if result.returncode != 0:
        raise RuntimeError("Scan failed. Review the output above.")
```

### Explanation

| Option | Purpose |
|---|---|
| `-sT` | TCP connect scan |
| `-Pn` | Skip the normal host-discovery decision |
| `-n` | Disable name resolution |
| `-p` | Select the practice port |

### Step 3: Scan Before Starting the Server

```python
scan_local()
```

Expected observation:

```text
closed
```

The port is reserved, but no service is listening.

### Step 4: Start a Temporary Web Server

```python
class PracticeHandler(http.server.BaseHTTPRequestHandler):
    def do_GET(self):
        body = (
            b"<html><head>"
            b"<title>My Nmap Practice Lab</title>"
            b"</head><body>"
            b"<h1>Hello from the Kaggle lab</h1>"
            b"</body></html>"
        )

        self.send_response(200)
        self.send_header("Content-Type", "text/html")
        self.send_header("Content-Length", str(len(body)))
        self.end_headers()
        self.wfile.write(body)

    def log_message(self, format, *args):
        pass

if server is None:
    reservation.close()
    reservation = None

    server = http.server.ThreadingHTTPServer(
        ("127.0.0.1", lab_port),
        PracticeHandler
    )

    server_thread = threading.Thread(
        target=server.serve_forever,
        daemon=True
    )

    server_thread.start()

print("Local web server is running.")
```

### Step 5: Confirm the Service Responds

```python
url = f"http://127.0.0.1:{lab_port}/"

with urllib.request.urlopen(url, timeout=5) as response:
    print("HTTP status:", response.status)
    print(response.read().decode())
```

Expected status:

```text
200
```

HTTP status 200 indicates a successful response.

### Step 6: Scan the Running Service

```python
scan_local("-sV", "--version-light")
```

Expected observation:

```text
open
```

`--version-light` requests a lighter set of service-detection probes.

Exact service and version labels may vary.

### Step 7: Retrieve the Page Title

```python
scan_local("--script", "http-title")
```

Look for:

```text
My Nmap Practice Lab
```

### Step 8: Save the Results

### Step 9: Stop the Web Server

```python
if server is not None:
    server.shutdown()
    server.server_close()
    server_thread.join(timeout=3)
    server = None

print("Web server stopped.")
```

## 22. Common Mistakes

### Mistake 1: Treating Every Open Port as a Vulnerability

A website normally needs a listening web service.

Investigate whether exposure is intended and appropriately secured.

### Mistake 2: Assuming the Default Scan Checks Everything

The default scan checks commonly used TCP ports.

It does not cover all TCP ports or UDP automatically.

### Mistake 3: Treating a Service Label as Confirmed Software

Without service detection, a label may reflect the port's conventional association.

### Mistake 4: Assuming -Pn Bypasses Firewalls

It changes discovery behavior, not firewall rules.

### Mistake 5: Assuming SYN Scans Are Invisible

Security monitoring can detect SYN scans.

### Mistake 6: Confusing Kaggle With Your Mac

Each environment has its own localhost.

### Mistake 7: Running Every Script

Scripts have different purposes and impacts. Select them deliberately.

## 23. Troubleshooting

| Problem | Possible explanation | Next step |
|---|---|---|
| `nmap: command not found` | Nmap is not installed | Install the executable |
| Installation fails | Internet or permissions unavailable | Use a permitted local environment |
| Raw-packet permission error | Scan requires unavailable privileges | Use TCP connect scanning |
| No open ports | No service listening or ports not selected | Check the service and scan scope |
| Host appears down | Discovery probes unanswered | Consider `-Pn` for a known authorized host |
| `open\|filtered` | Responses are ambiguous | Investigate further; do not assume open |
| Local browser cannot reach Kaggle service | Different localhost environments | Use an HTTP request inside the notebook |
| Python variable not defined | Cells ran out of order | Run setup cells first |
| Address already in use | Port was occupied | Rerun setup to choose another port |
