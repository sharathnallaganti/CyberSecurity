# Access Control in Cybersecurity

## Overview

Access control determines **who can access a resource, what actions they can perform, and under which conditions**.

Resources include files, applications, APIs, database records, cloud services, networks, and AI tools.

For example, a SOC analyst may read security alerts but may not delete audit logs or create administrator accounts.

## Learning Objectives

After completing this guide, you should be able to:

- Explain authentication and authorization.
- Compare the main access-control methods.
- Understand the technologies used to enforce access.
- Implement role, ownership, and tenant checks.
- Recognize common access-control vulnerabilities.
- Test access control using practical tools.
- Apply access control to LLM, RAG, and AI-agent systems.

# 1. Fundamental Concepts

## 1.1 Identification

Identification is claiming an identity.

Examples:

- Entering a username.
- Providing an email address.
- Specifying a service-account identifier.

Identification alone does not prove that the claim is legitimate.

## 1.2 Authentication

Authentication verifies the claimed identity.

Common methods include:

- Passwords.
- Passkeys.
- Hardware security keys.
- One-time passwords.
- Client certificates.
- Multifactor authentication.

**Example:** A user enters a password and completes an MFA challenge.

## 1.3 Authorization

Authorization determines which actions an authenticated identity may perform.

**Example:** A user successfully logs in but cannot access the administrator dashboard because they do not have the required permission.

Authentication and authorization are separate decisions.

## 1.4 Accountability

Accountability allows an organization to trace actions to identities.

Common mechanisms include:

- Audit logs.
- Access logs.
- Permission-change records.
- Administrative activity records.

An audit event should identify the actor, action, resource, time, and outcome.

## 1.5 Components of an Access Decision

| Component | Meaning | Example |
|---|---|---|
| Subject | Identity requesting access | User or service account |
| Resource | Protected object | Incident report |
| Action | Requested operation | Read or update |
| Context | Relevant conditions | Tenant, device, time |
| Decision | Policy outcome | Allow or deny |

Example:

```text
Subject: Sharath
Resource: Incident report 103
Action: Read
Context: Same company and assigned analyst
Decision: Allow
```

# 2. Access-Control Methods

## 2.1 Discretionary Access Control — DAC

DAC allows resource owners to assign permissions.

**Example:** A file owner gives another user read access.

### Technologies

- File ownership.
- Linux permissions.
- Access-control lists.
- File-sharing permissions.

### Limitation

Owners may share resources too broadly or assign incorrect permissions.

## 2.2 Mandatory Access Control — MAC

MAC uses centrally enforced policies that users cannot freely override.

Policies may use security labels, classifications, or process restrictions.

**Example:** A process is prevented from reading a protected file even when ordinary file permissions would otherwise allow access.

### Technologies

- SELinux.
- AppArmor.
- Security-label systems.

MAC policies can control both confidentiality and integrity.

## 2.3 Role-Based Access Control — RBAC

RBAC assigns permissions to roles and assigns roles to users.

| Role | Example Permissions |
|---|---|
| Viewer | Read assigned reports |
| SOC analyst | Read and investigate alerts |
| SOC manager | Assign incidents |
| Tenant administrator | Manage users within one tenant |

### Advantages

- Easier permission administration.
- Clear mapping between job responsibilities and access.
- Reusable permission groups.

### Limitation

Roles may become too broad, or the system may accumulate too many roles.

## 2.4 Attribute-Based Access Control — ABAC

ABAC evaluates attributes of the subject, resource, action, and environment.

Example rule:

```text
Allow access when:

user.department == resource.department
AND user.tenant_id == resource.tenant_id
AND device.managed == True
```

### Common Attributes

- Department.
- Tenant.
- Resource classification.
- Device posture.
- Location.
- Time.

Attributes must come from trusted sources.

## 2.5 Relationship-Based Access Control — ReBAC

ReBAC bases permissions on relationships.

Examples:

- A document owner can edit the document.
- A project member can view project resources.
- An assigned analyst can investigate an incident.

## 2.6 Combining Methods

Real systems often combine several methods.

Example SOC application:

- RBAC determines whether the user can investigate incidents.
- ABAC enforces company boundaries.
- Relationship checks determine whether the user is assigned to the incident.

# 3. Core Security Principles

## 3.1 Least Privilege

Grant only the permissions required for a task.

**Example:** A reporting service needs database read access but does not need permission to delete tables.

## 3.2 Deny by Default

Reject access unless a policy explicitly permits it.

Unknown roles, unsupported actions, and missing permissions should not accidentally receive access.

## 3.3 Need to Know

Users should access information relevant to their responsibilities.

A valid employee account does not imply permission to view every employee record.

## 3.4 Separation of Duties

Divide sensitive activities between different identities.

**Example:** One employee requests a payment, and another approves it.

## 3.5 Check Every Request

Enforce authorization for every relevant operation.

Do not assume that access to a page establishes permission for all requests made from that page.

## 3.6 Access Lifecycle Management

Access should be:

1. Provisioned when needed.
2. Updated when responsibilities change.
3. Reviewed periodically.
4. Revoked when no longer required.

# 4. Access-Control Architecture

## 4.1 Policy Enforcement Point — PEP

The PEP intercepts an operation and enforces the decision.

Examples:

- API middleware.
- Endpoint dependencies.
- API gateways.
- Operating-system permission checks.

## 4.2 Policy Decision Point — PDP

The PDP evaluates the access policy.

It answers:

```text
Can this subject perform this action on this resource?
```

A policy engine such as Open Policy Agent can perform this responsibility.

## 4.3 Policy Information Point — PIP

The PIP supplies attributes needed for the decision.

Examples:

- User role.
- Tenant membership.
- Document ownership.
- Device status.

## 4.4 Policy Administration Point — PAP

The PAP manages policy definitions.

These responsibilities may exist inside one application or across separate services.

# 6. Authentication and Authorization Protocols

## 6.1 OAuth 2.0

OAuth 2.0 supports delegated authorization.

An application receives limited access to an API without necessarily receiving the user's password.

Example scope:

```text
reports:read
```

OAuth 2.0 is not itself a user-login protocol.

## 6.2 OpenID Connect — OIDC

OIDC adds an authentication layer to OAuth 2.0.

It allows an application to establish who signed in.

- An ID token communicates authentication information to the client.
- An access token is used to access an API.

## 6.3 SAML

SAML exchanges assertions between an identity provider and an application.

It is commonly used for enterprise single sign-on.

## 6.4 LDAP

LDAP accesses directory information.

Examples:

- User accounts.
- Groups.
- Organizational attributes.

LDAP is not a complete application authorization system.

## 6.5 Kerberos

Kerberos uses tickets to authenticate users and services.

Resource permissions still need to be enforced after authentication.

## 6.6 RADIUS

RADIUS supports authentication, authorization, and accounting for network access.

It is often used with network admission systems.

## 6.7 JSON Web Tokens — JWT

JWT is a token format.

A JWT may carry claims such as:

```json
{
  "sub": "user-101",
  "tenant_id": "company-a",
  "scope": "reports:read"
}
```

A JWT is not an access-control model.

A signed JWT is generally readable. Signing does not encrypt its payload.

### JWT Validation

The API should verify:

- Signature.
- Permitted signing algorithm.
- Trusted issuer.
- Intended audience.
- Expiration.
- Other required claims.

Decoding a token without validating it does not establish trust.

# 7. Application and API Access Control

## 7.1 Function-Level Authorization

Checks whether the user may perform an operation.

Example:

```text
Can this user delete reports?
```

## 7.2 Object-Level Authorization

Checks whether the user may access a particular object.

Example:

```text
Can this user read report 103?
```

## 7.3 Tenant Isolation

A tenant represents a customer or organization in a shared application.

Tenant isolation prevents one customer from accessing another customer's resources.

A tenant administrator should not automatically receive global administrator privileges.

## 7.4 Request Processing

For this request:

```http
GET /reports/103
Authorization: Bearer <access-token>
```

The server should:

1. Validate the credential.
2. Establish a trusted identity.
3. Check the requested operation.
4. Check object and tenant permissions.
5. Perform the permitted operation.
6. Record relevant audit information.

Never trust caller-supplied role, owner, or tenant fields as proof of permission.

# 8. Practical Python Lab

## 8.1 Lab Objective

Implement:

- Role permissions.
- Ownership checks.
- Tenant isolation.
- Deny-by-default behavior.

The following code uses synthetic identities that are assumed to have already been authenticated.

Paste it into a Kaggle **Code cell** to execute it.

## 8.2 Define Users, Reports, and Permissions

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class User:
    id: str
    tenant_id: str
    role: str

@dataclass(frozen=True)
class Report:
    id: int
    tenant_id: str
    owner_id: str

ROLE_PERMISSIONS = {
    "viewer": {"reports:read"},
    "analyst": {"reports:read", "reports:update"},
    "tenant_admin": {"reports:read", "reports:update"},
}
```

## 8.3 Implement Authorization

```python
def authorize(user, action, report):
    if user is None:
        return False

    permissions = ROLE_PERMISSIONS.get(user.role, set())

    if action not in permissions:
        return False

    if user.tenant_id != report.tenant_id:
        return False

    if user.role != "tenant_admin" and user.id != report.owner_id:
        return False

    return True
```

### Explanation

The function rejects access when:

- The user is unauthenticated.
- The role lacks the requested permission.
- The tenant does not match.
- The user does not own the report and is not its tenant administrator.

The tenant check runs before the administrator exception, preventing cross-tenant access.

## 8.4 Create Test Identities

```python
alice = User("alice", "company-a", "analyst")
bob = User("bob", "company-a", "viewer")
admin = User("admin-a", "company-a", "tenant_admin")
outsider = User("admin-b", "company-b", "tenant_admin")

alice_report = Report(101, "company-a", "alice")
bob_report = Report(102, "company-a", "bob")
```

## 8.5 Test the Policy

```python
assert authorize(alice, "reports:read", alice_report)
assert not authorize(alice, "reports:read", bob_report)
assert not authorize(bob, "reports:update", bob_report)
assert authorize(admin, "reports:read", bob_report)
assert not authorize(outsider, "reports:read", alice_report)
assert not authorize(None, "reports:read", alice_report)
assert not authorize(alice, "reports:delete", alice_report)

print("All authorization checks passed.")
```

These checks validate the demonstration policy, not a production authentication system.

# 9. FastAPI Implementation Pattern

## 9.1 Endpoint Authorization

The following is an illustrative pattern. Authentication and repository functions must be implemented separately.

```python
@app.get("/reports/{report_id}")
def read_report(
    report_id: int,
    user=Depends(get_current_user)
):
    if "reports:read" not in user.permissions:
        raise HTTPException(
            status_code=403,
            detail="Forbidden"
        )

    report = repository.find_report(
        report_id=report_id,
        tenant_id=user.tenant_id,
        owner_id=user.id
    )

    if report is None:
        raise HTTPException(
            status_code=404,
            detail="Report not found"
        )

    return report
```

This endpoint uses an owner-only policy.

`get_current_user` must validate credentials and return trusted identity information.

The repository query must enforce the ownership and tenant conditions.

## 9.2 HTTP Responses

| Status | Typical Meaning |
|---|---|
| 401 | Missing or invalid authentication |
| 403 | Authenticated caller lacks permission |
| 404 | Resource does not exist or its existence is concealed |

A response code alone does not prove that authorization is correct. Check whether restricted data or operations were exposed.

# 10. Linux Access Control

## 10.1 Basic Permission Bits

| Permission | Numeric Value | File Meaning |
|---|---|---|
| Read | 4 | Read contents |
| Write | 2 | Modify contents |
| Execute | 1 | Execute the file |

Permissions apply to:

- Owner.
- Group.
- Others.

## 10.2 chmod

```bash
chmod 640 report.txt
```

| Identity | Permission |
|---|---|
| Owner | Read and write |
| Group | Read |
| Others | None |

## 10.3 chown

```bash
sudo chown sharath:soc report.txt
```

This changes the owner to `sharath` and the group to `soc`.

## 10.4 Access-Control Lists

```bash
setfacl -m u:alice:r report.txt
getfacl report.txt
```

The first command adds read permission for Alice.

The second displays the ACL.

The ACL mask can constrain effective permissions.

## 10.5 Directory Permissions

Directory permissions behave differently from file permissions:

- Read allows listing names.
- Execute allows traversal.
- Write generally permits changing directory entries when traversal is also available.

Deleting a file primarily depends on parent-directory permissions, with additional rules such as the sticky bit.

## 10.6 Privileged Commands and Process Controls

| Tool | Purpose |
|---|---|
| sudo | Execute permitted commands as another identity |
| sudo -l | Inspect permitted privileged commands |
| SELinux | Enforce mandatory security policies |
| AppArmor | Restrict process behavior through profiles |

These are Linux reference examples. POSIX ACL commands may require additional packages.

# 11. Network Access Control

## 11.1 Firewalls and Network ACLs

Firewalls control connections using configured rules.

Example:

```text
Allow the API server to reach PostgreSQL on TCP 5432.
Reject direct internet connections to PostgreSQL.
```

This does not replace database or application authorization.

## 11.2 Network Segmentation

VLANs and network segments separate groups of systems.

Examples:

- Employee devices.
- Database servers.
- Administrative systems.
- Guest devices.

Segmentation reduces reachable attack paths but does not establish user-level permissions.

## 11.3 Network Admission

802.1X and RADIUS can authenticate identities before allowing network access.

Network Access Control systems may also consider device posture.

## 11.4 VPN Access

A VPN provides a protected connection.

VPN membership does not imply unrestricted access to every internal resource.

# 12. AWS IAM Access Control

## 12.1 IAM Policy Components

| Field | Purpose |
|---|---|
| Effect | Allow or Deny |
| Action | Permitted or denied operation |
| Resource | Target resource |
| Condition | Additional requirements |

## 12.2 Example S3 Policy

```json
{
  "Version": "2012-10-17",
  "Statement": [
    {
      "Effect": "Allow",
      "Action": "s3:GetObject",
      "Resource": "arn:aws:s3:::example-security-bucket/reports/*"
    }
  ]
}
```

This statement grants object retrieval under the specified prefix.

It does not itself grant upload, deletion, or bucket listing.

Effective permissions also depend on other applicable policies.

## 12.3 Important IAM Concepts

- Identity policies grant permissions to identities.
- Resource policies grant access through supported resources.
- Role trust policies define who may assume a role.
- Role permission policies define what the assumed role may do.
- Temporary credentials reduce reliance on long-lived keys.
- Permissions boundaries and service control policies constrain permissions; they do not grant permissions by themselves.
- An applicable explicit deny overrides an allow.

## 12.4 Useful Tools

- AWS IAM Access Analyzer.
- IAM policy simulation.
- AWS CloudTrail.

Interpret results in the context of relevant resource and organization policies.

# 13. Database Access Control

## 13.1 Table Permissions

PostgreSQL example:

```sql
GRANT SELECT ON alerts TO soc_reader;

REVOKE INSERT, UPDATE, DELETE
ON alerts FROM soc_reader;
```

The table and role must already exist.

Access may still be inherited from other roles or granted through another path.

## 13.2 Row-Level Security

Row-level security controls which individual rows a database role may access.

Example:

```text
Company A can read Company A records.
Company B can read Company B records.
```

RLS supplements table permissions.

A shared application database login does not automatically identify each application user.

## 13.3 Database Role Design

Use separate roles for:

- Application runtime.
- Reporting.
- Migrations.
- Database administration.

The application should not normally connect as a database superuser.

In PostgreSQL, superusers and BYPASSRLS roles bypass row-level security. Table owners normally bypass it unless FORCE ROW LEVEL SECURITY applies.

# 14. Kubernetes Access Control

## 14.1 RBAC Components

| Component | Purpose |
|---|---|
| Role | Defines permissions within a namespace |
| ClusterRole | Defines cluster-scoped or reusable permissions |
| RoleBinding | Grants permissions within a namespace |
| ClusterRoleBinding | Grants permissions across the cluster |
| ServiceAccount | Represents a workload identity |

A binding grants permissions to an identity. It does not create a user's login credential.

## 14.2 Inspect Permissions

```bash
kubectl auth can-i get pods --namespace development
```

This checks whether the current identity can retrieve pods in the specified namespace.

## 14.3 API Access Versus Network Access

Kubernetes RBAC controls operations on the Kubernetes API.

NetworkPolicy controls supported pod networking behavior.

Avoid unnecessary access to Secrets, broad wildcard grants, and powerful workload-management permissions.

# 15. Common Access-Control Vulnerabilities

## 15.1 Horizontal Privilege Escalation

A user accesses another user's resources at the same privilege level.

**Example:** Alice reads Bob's private report.

## 15.2 Vertical Privilege Escalation

A user performs an operation reserved for a more privileged role.

**Example:** A viewer calls an administrator-only endpoint.

## 15.3 IDOR and BOLA

IDOR means Insecure Direct Object Reference.

BOLA means Broken Object Level Authorization.

Example request:

```http
GET /reports/101
```

If changing the identifier to another user's report exposes that report without authorization, an object-level failure exists.

Object identifiers can be numbers, UUIDs, filenames, or other references.

## 15.4 Broken Function-Level Authorization

An endpoint fails to check whether the caller may perform its operation.

**Example:** Any authenticated user can call a delete endpoint.

## 15.5 Broken Property-Level Authorization

The application allows access to fields the caller should not read or modify.

**Example:** A profile-update request accepts an unauthorized `role` field.

## 15.6 Cross-Tenant Access

The application fails to enforce customer boundaries.

**Example:** Company A can export Company B records.

## 15.7 Excessive Service Permissions

A workload receives permissions beyond its purpose.

**Example:** A reporting service can delete storage objects.

# 16. Access-Control Testing

## 16.1 Testing Tools

| Tool | Use |
|---|---|
| Burp Suite Repeater | Replay requests with different test-account sessions |
| Postman | Send and organize API requests |
| curl | Send requests from a terminal |
| pytest | Automate authorization checks |
| getfacl | Inspect Linux ACLs |
| kubectl auth can-i | Inspect Kubernetes API permissions |
| Audit logs and SIEM | Investigate decisions and changes |

## 16.2 Two-Account Testing Method

Use a local lab or an explicitly authorized target.

1. Create Alice and Bob test accounts.
2. Give each account a private resource.
3. Capture Alice's normal request.
4. Replace the resource identifier with Bob's identifier.
5. Keep Alice's credential.
6. Verify that access is rejected.
7. Repeat for permitted test operations.
8. Fix the server-side policy and retest.

A successful HTTP response is not enough to establish a vulnerability. Verify whether restricted data or an unauthorized action was exposed.

## 16.3 Test Matrix

| Scenario | Expected Result |
|---|---|
| Owner reads own resource | Allow |
| User reads another user's private resource | Deny |
| User accesses another tenant | Deny |
| Viewer calls administrator operation | Deny |
| Credential is missing or invalid | Reject authentication |
| Caller supplies a forged role or tenant | No privilege gain |
| Permission is revoked | Access follows the revocation policy |
| User searches or exports records | Only authorized records returned |

Test downloads, attachments, bulk actions, search results, and exports as well as ordinary detail pages.

# 18. Access Control for LLM, RAG, and AI Agents

## 18.1 RAG Document Authorization

A RAG system should retrieve only documents the authenticated user is allowed to access.

Recommended sequence:

1. Authenticate the user.
2. Establish trusted tenant and document permissions.
3. Apply authorization filters during retrieval.
4. Verify access before adding content to model context.
5. Protect citations, previews, downloads, and caches.

Filtering only the final answer is too late if restricted information has already entered the model context.

## 18.2 AI Tool Authorization

A model-generated tool call must pass server-side authorization.

Example:

```text
Alice asks the assistant to read Bob's private report.

The model proposes read_report(102).

The backend checks Alice's identity and denies access.
```

The model cannot grant permissions through instructions.

## 18.3 Trusted Identity

The backend should obtain identity from the authenticated application context.

It should not trust model-generated arguments such as:

```json
{
  "user_id": "administrator",
  "tenant_id": "another-company"
}
```

## 18.4 Least Privilege for Agents

Give each agent or tool only the permissions it needs.

Examples:

- A search tool can retrieve permitted documents.
- A summarization tool cannot delete files.
- A read-only assistant cannot update database records.
- A tenant-specific tool cannot access another tenant.

Conversation history and caches also need authorization controls.

# 19. Practice Questions

## Question 1: Does MFA solve broken access control?

No. MFA strengthens authentication. Authorization must still check permissions.

## Question 2: Why is a read scope insufficient for private objects?

The scope permits a type of action. Ownership, tenant, or relationship rules determine which objects are accessible.

## Question 3: Which model uses department and device attributes?

ABAC.

## Question 4: Does signing a JWT hide its contents?

No. Signing protects integrity and authenticity when properly validated. It does not provide encryption.

## Question 5: Does hiding an administrator button protect an endpoint?

No. Users can send requests directly. The server must enforce authorization.

## Question 6: Does a Kubernetes RoleBinding create a user?

No. It grants permissions to an identity established separately.

## Question 7: Can an LLM prompt grant permission to a tool?

No. The tool backend must authorize the authenticated caller independently.
