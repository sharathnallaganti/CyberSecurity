# Networking for Cybersecurity




## 1.1 What Is a Network?

A network connects devices so they can exchange data.

Connections can use:

- Copper cables.
- Fiber-optic cables.
- Wireless radio signals.

**Example:** A laptop communicates with a web server using network protocols.

## 1.2 What Is a Protocol?

A protocol defines rules for communication.

It specifies how messages are structured, exchanged, and interpreted.

| Protocol | Main Purpose |
|---|---|
| IP | Addressing and packet delivery across networks |
| TCP | Reliable, ordered byte-stream delivery |
| UDP | Datagram transport without built-in delivery guarantees |
| DNS | Resolving names and other DNS information |
| HTTP | Application requests and responses |
| TLS | Protecting communication and authenticating peers |

A protocol performs a particular job.

**Example:** TCP supports reliable delivery. It does not determine whether a user may read a private report.

## 1.3 What Is a Packet?

An IP packet contains a header and a payload.

The header includes information such as:

- Source IP address.
- Destination IP address.
- Upper-layer protocol.
- Lifetime information, such as TTL.

## 1.4 Host, Node, and Link

| Term | Meaning |
|---|---|
| Host | Device that sends or receives application traffic |
| Node | Device participating in the network |
| Link | Connection between network nodes |

## 1.5 Network Performance Terms

| Term | Meaning |
|---|---|
| Bandwidth | Available or nominal capacity of a connection |
| Throughput | Actual rate of successful data transfer |
| Latency | Time taken for communication or delivery |
| Jitter | Variation in delay |
| Packet loss | Packets that fail to reach their destination |

Latency measurements may describe one-way delay or round-trip time.

**Example:** A high-bandwidth connection may still feel slow when round-trip latency is high.

Voice and video applications can be affected by jitter and packet loss.

## 2.2 Virtual Private Network — VPN

A VPN creates a logical private connection over another network.

Remote-access VPNs commonly encrypt traffic between a user's device and a VPN gateway.

A VPN is not a network-size category.

**Example:** A remote employee connects to company resources through an encrypted tunnel.

## 2.3 Networks by Organization

| Type | Description |
|---|---|
| Client-server | Clients request resources from servers |
| Peer-to-peer | Peers exchange resources directly |
| Intranet | Network or services intended for internal use |
| Extranet | Selected services made available to external partners |
| Internet | Interconnected networks using the Internet protocol suite |

## 2.4 Network Topologies

| Topology | Structure | Example |
|---|---|---|
| Bus | Devices share a common medium | Historical Ethernet designs |
| Star | Devices connect to a central device | Office devices connected to a switch |
| Mesh | Multiple connections provide alternative paths | Router networks |
| Hybrid | Combination of structures | Enterprise networks |

## 2.5 Traffic Delivery Types

| Type | Meaning |
|---|---|
| Unicast | Traffic addressed to one destination |
| Broadcast | Traffic addressed to all devices in a broadcast domain |
| Multicast | Traffic addressed to a receiver group |
| Anycast | Multiple destinations share an address; routing selects one |

Anycast selects a destination according to routing policy and topology. It does not necessarily select the geographically nearest server.

IPv6 uses multicast rather than broadcast.

# 3. OSI and TCP/IP Models

## 3.1 OSI Layers

| Layer | Name | Main Responsibility | Examples |
|---|---|---|---|
| 7 | Application | Application communication | HTTP, DNS, SMTP |
| 6 | Presentation | Data representation and transformation | Encoding and serialization concepts |
| 5 | Session | Managing communication sessions | Session-management concepts |
| 4 | Transport | End-to-end transport | TCP, UDP |
| 3 | Network | Addressing and routing | IPv4, IPv6, ICMP |
| 2 | Data Link | Local-link communication | Ethernet, Wi-Fi |
| 1 | Physical | Transmission of signals | Copper, fiber, radio |

The OSI model is a conceptual model. Real protocols do not always fit neatly into one layer.

## 3.2 Mapping to TCP/IP

| OSI Layers | TCP/IP Layer | Examples |
|---|---|---|
| Application, Presentation, Session | Application | HTTP, DNS, SSH, SMTP |
| Transport | Transport | TCP, UDP |
| Network | Internet | IPv4, IPv6, ICMP |
| Data Link, Physical | Link | Ethernet, Wi-Fi |

ARP operates at the boundary between IPv4 addressing and local-link delivery.

## 3.3 Encapsulation

Encapsulation wraps application data in headers needed for delivery.

| Stage | Information Added | Common Name |
|---|---|---|
| Application | Application message structure | Data |
| Transport | TCP or UDP header, including ports | Segment or datagram |
| Internet | IP header | Packet |
| Link | Link header and, where applicable, trailer | Frame |

At a router, the incoming link frame is removed and a new frame is constructed for the outgoing link.

The IP packet is forwarded, but it is not completely unchanged: fields such as TTL or Hop Limit change, and NAT may rewrite addresses.

# 4. Network Devices

## 4.1 Device Comparison

| Device | Main Job |
|---|---|
| Hub | Repeats received physical signals to other ports |
| Layer 2 switch | Forwards frames using MAC-address information |
| Wireless access point | Connects wireless clients to a network |
| Router | Forwards packets between IP networks |
| Firewall | Enforces traffic policy |
| Proxy | Relays communication on behalf of a client or server |
| Load balancer | Distributes connections or requests across backends |

A Layer 3 switch can also route.

A router chooses the next hop using IP routing information, then uses the outgoing link's addressing to transmit the packet.

## 4.2 MAC Address Versus IP Address

| Address | Main Role |
|---|---|
| MAC address | Local-link delivery on technologies such as Ethernet |
| IP address | Delivery across IP networks |

**Example:** When reaching a remote server, a laptop usually sends the Ethernet frame to its gateway's MAC address while retaining the remote server's IP address as the packet destination.

# 5. IP Addressing and Subnetting

## 5.1 IPv4 Address

An IPv4 address contains 32 bits.

Example:

```text
192.168.10.77
```

It is commonly written as four decimal octets.

## 5.2 Prefix Length

A prefix indicates how many bits identify the network.

Example:

```text
192.168.10.77/26
```

The first 26 bits identify the network.

The remaining 6 bits identify addresses within the subnet.

## 5.3 Common IPv4 Subnets

| Prefix | Subnet Mask | Total Addresses | Conventional Usable Host Addresses |
|---|---|---:|---:|
| /30 | 255.255.255.252 | 4 | 2 |
| /28 | 255.255.255.240 | 16 | 14 |
| /26 | 255.255.255.192 | 64 | 62 |
| /24 | 255.255.255.0 | 256 | 254 |
| /16 | 255.255.0.0 | 65,536 | 65,534 |

For these conventional subnets:

```text
Total addresses = 2^(32 - prefix length)

Usable host addresses = total addresses - 2
```

The subtraction accounts for the network and broadcast addresses.

There are exceptions:

- A /31 can use both addresses on an IPv4 point-to-point link.
- A /32 represents one address.
- Cloud platforms may reserve additional addresses.

## 5.4 Important IPv4 Ranges

| Range | Purpose |
|---|---|
| 10.0.0.0/8 | Private addressing |
| 172.16.0.0/12 | Private addressing |
| 192.168.0.0/16 | Private addressing |
| 127.0.0.0/8 | Loopback |
| 169.254.0.0/16 | IPv4 link-local |
| 224.0.0.0/4 | Multicast |

The private 172 range extends from `172.16.0.0` through `172.31.255.255`.

A link-local address may indicate that a device could not obtain a DHCP lease, but it does not prove the exact cause.

## 5.5 Determining Whether a Destination Is Local

A host compares the destination against its configured subnet and routing information.

Example:

```text
192.168.1.10/24
192.168.1.20/24
```

These addresses are in the same subnet.

Example:

```text
192.168.1.10/24
192.168.2.10/24
```

These addresses are in different subnets.

A host normally sends traffic for a remote subnet to an appropriate gateway.

## 5.6 Worked Subnetting Example

Given:

```text
192.168.10.77/26
```

Calculation:

- Host bits: 6.
- Addresses per block: 64.
- Relevant block: 64–127.
- Network address: `192.168.10.64`.
- Broadcast address: `192.168.10.127`.
- Conventional usable range: `192.168.10.65–192.168.10.126`.

# 6. Local Delivery Protocols

## 6.1 Address Resolution Protocol — ARP

ARP resolves an IPv4 address to a link-layer address on the local network.

Example:

```text
Who has 192.168.1.1?
```

A normal ARP request is broadcast, and a reply is commonly unicast.

ARP information is cached.

### Security Concern

ARP does not provide built-in authentication.

Forged ARP information can redirect traffic on a local network.

ARP broadcasts are not ordinarily forwarded by routers. Proxy ARP is a separate behavior where a device answers on behalf of another address.

IPv6 uses Neighbor Discovery rather than ARP.

## 6.2 Dynamic Host Configuration Protocol — DHCP

DHCP can provide:

- IP address.
- Subnet mask.
- Default gateway.
- DNS servers.
- Lease duration.

### Initial DHCP Exchange

| Step | Message | Purpose |
|---|---|---|
| D | Discover | Client searches for DHCP servers |
| O | Offer | Server proposes configuration |
| R | Request | Client requests an offered configuration |
| A | Acknowledge | Server confirms the lease |

This sequence is commonly called DORA.

DHCP for IPv4 uses:

- UDP 67 for the server.
- UDP 68 for the client.

DHCP relay agents can forward DHCP communication between subnets.

## 6.3 Internet Control Message Protocol — ICMP

ICMP communicates network-control and error information.

### Common ICMPv4 Types

| Type | Meaning |
|---|---|
| 8 | Echo request |
| 0 | Echo reply |
| 3 | Destination unreachable |
| 11 | Time exceeded |

Ping commonly uses echo requests and replies.

Traceroute commonly relies on time-exceeded responses.

### Path MTU Discovery

Path MTU discovery needs appropriate error messages:

- IPv4 uses fragmentation-needed information.
- IPv6 uses ICMPv6 Packet Too Big messages.

Blocking all ICMP can break useful diagnostics and network behavior.

# 7. NAT and Port Address Translation

## 7.1 Network Address Translation — NAT

NAT rewrites IP addresses as traffic passes through a device.

## 7.2 Port Address Translation — PAT

PAT also translates transport-layer ports.

It allows multiple internal connections to share a public IP address.

Example:

```text
Internal connection:
192.168.1.10:51500

Translated connection:
Public-IP:62001
```

## 7.3 NAT Versus Firewall

NAT performs translation.

A firewall enforces traffic policy.

NAT may affect unsolicited inbound reachability, but it is not a substitute for explicit security policy.

Port forwarding can make an internal service reachable from outside.

# 8. TCP and UDP

## 8.1 Comparison

| Feature | TCP | UDP |
|---|---|---|
| Connection setup | Three-way handshake | No transport handshake |
| Delivery | Reliable, ordered byte stream | No built-in delivery or ordering guarantee |
| Data boundaries | Does not preserve application message boundaries | Preserves datagram boundaries |
| Retransmission | Built in | Not built in |
| Typical uses | SSH, SMTP, HTTP/1.1 and HTTP/2 | DNS, DHCP, NTP, real-time traffic |

Applications can implement reliability over UDP.

QUIC is an important example: it runs over UDP and supplies secure, reliable transport features.

## 8.2 TCP Three-Way Handshake

1. Client sends SYN.
2. Server sends SYN-ACK.
3. Client sends ACK.

| Message | Purpose |
|---|---|
| SYN | Starts connection establishment |
| SYN-ACK | Acknowledges the client's SYN and sends the server's SYN |
| ACK | Acknowledges the server's SYN |

## 8.3 Common TCP Flags

| Flag | Purpose |
|---|---|
| SYN | Establish connection |
| ACK | Acknowledge received sequence information |
| FIN | Request orderly closure of one sending direction |
| RST | Reset a connection |

TCP can acknowledge ranges of bytes and retransmit missing data.

## 8.4 SYN Flood

A SYN flood attempts to exhaust resources associated with connection establishment.

SYN cookies can reduce the need to retain state for incomplete handshakes.

## 8.5 Spoofing Considerations

Completing a TCP handshake generally requires receiving or correctly predicting the server's response.

UDP has no transport handshake, which can make source-address spoofing easier in some scenarios.

Neither TCP nor UDP independently establishes a trustworthy user identity.

# 9. Common Ports

## 9.1 Port Reference Table

| Port | Transport | Common Service |
|---|---|---|
| 22 | TCP | SSH |
| 23 | TCP | Telnet |
| 25 | TCP | SMTP server-to-server communication |
| 53 | UDP and TCP | DNS |
| 67/68 | UDP | DHCPv4 |
| 80 | TCP | HTTP |
| 443 | TCP | HTTPS using HTTP/1.1 or HTTP/2 |
| 443 | UDP | HTTP/3 over QUIC |
| 445 | TCP | SMB |
| 3389 | TCP and UDP | RDP |

A port number is a convention, not proof of the application protocol.

## 9.2 Security Considerations

- Restrict administrative services such as SSH and RDP.
- Telnet transmits information without transport encryption.
- SMB exposure can contribute to lateral movement.
- DNS traffic may require investigation for tunneling or amplification.
- Encrypted traffic can still carry malicious activity.

# 10. DNS

## 10.1 Domain Name System

DNS provides a distributed naming system.

It can resolve a hostname into an IP address and return other record types.

## 10.2 Typical Lookup

For an uncached lookup:

1. The client asks a recursive resolver.
2. The resolver contacts a root server.
3. The root server refers it to a top-level-domain server.
4. The TLD server refers it to an authoritative server.
5. The authoritative server returns the requested information.
6. The resolver returns the result to the client.

Caches can eliminate some of these steps.

## 10.3 Common DNS Records

| Record | Purpose |
|---|---|
| A | IPv4 address |
| AAAA | IPv6 address |
| CNAME | Alias to another name |
| MX | Mail-exchange information |
| NS | Name-server information |
| TXT | Text information, including uses such as SPF |
| PTR | Reverse lookup |

## 10.4 DNS Security Concerns

Examples include:

- DNS spoofing.
- Cache poisoning.
- DNS tunneling.
- Reflection and amplification.
- Malicious domains.

DNSSEC provides authenticity and integrity for signed DNS data. It does not encrypt DNS queries.

# 11. HTTP

## 11.1 HTTP Request Structure

The following uses HTTP/1.1 notation:

```http
POST /api/orders HTTP/1.1
Host: shop.example
Content-Type: application/json
Cookie: session=example-value

{"sku": "cable", "qty": 2}
```

| Component | Meaning |
|---|---|
| POST | Request method |
| /api/orders | Target path |
| Host | Requested host |
| Content-Type | Body format |
| Cookie | Client-supplied cookie |
| JSON body | Submitted application data |

Modern HTTP versions use different on-wire representations.

## 11.2 HTTP Methods

| Method | Standard Purpose |
|---|---|
| GET | Retrieve a representation |
| POST | Submit data for processing |
| PUT | Create or replace the target resource's state |
| PATCH | Apply partial modifications |
| DELETE | Request removal of the target resource |
| HEAD | Request GET-like headers without response content |
| OPTIONS | Request communication options |

The application implementation determines actual behavior.

A method name does not prove that an operation is correctly authorized.

## 11.3 HTTP Response Structure

```http
HTTP/1.1 201 Created
Content-Type: application/json

{"order_id": 501, "status": "created"}
```

## 11.4 Common Status Codes

| Code | General Meaning |
|---|---|
| 200 | Request succeeded |
| 201 | Resource created |
| 204 | Success without response content |
| 301 | Permanent redirect |
| 302 | Temporary redirect |
| 400 | Bad request |
| 401 | Authentication required or not accepted |
| 403 | Request forbidden |
| 404 | Resource not found or existence concealed |
| 409 | Conflict with current resource state |
| 429 | Too many requests |
| 500 | Internal server error |

Read response content and verify the resulting application state.

A `200` response alone does not prove that the intended operation succeeded.

# 12. TLS and HTTPS

## 12.1 Transport Layer Security — TLS

TLS protects communication through:

- Confidentiality.
- Integrity.
- Peer authentication.

In typical HTTPS usage, the client authenticates the server using its certificate.

Mutual TLS can also authenticate the client.

## 12.2 Certificate Validation

A client typically checks:

- Whether the requested hostname matches.
- Whether the certificate chain is trusted.
- Whether the certificate is within its validity period.
- Other applicable certificate requirements.

A valid certificate does not prove that a website is harmless.

## 12.3 TLS Versus Authorization

TLS protects communication and can authenticate peers.

Application authorization still determines whether a user may read a report, modify an account, or call an administrative endpoint.

## 12.4 Common TLS Problems

- Expired certificate.
- Hostname mismatch.
- Missing or untrusted certificate chain.
- Incorrect device clock.
- Incompatible protocol configuration.

# 13. Tracing a Web Request

## 13.1 Typical HTTPS Request over TCP

When a browser visits an HTTPS website:

1. It resolves the hostname, unless an appropriate cached result exists.
2. It determines the route to the destination.
3. It obtains local next-hop addressing if needed.
4. It establishes a TCP connection.
5. It negotiates TLS.
6. It sends the HTTP request.
7. It receives the response.

Connections and cached information can be reused.

HTTP/3 uses QUIC over UDP rather than a TCP connection.

## 13.2 Questions for an Investigator

- Did DNS return the expected result?
- Is there a route to the destination?
- Can the next hop be reached?
- Did transport connection establishment succeed?
- Did TLS validation succeed?
- Did the application return an error?
- Did authorization reject the user?

# 14. Network Security Controls

## 14.1 Control Comparison

| Control | Primary Purpose | Important Limitation |
|---|---|---|
| Stateless firewall | Evaluate packets against rules | Does not track connection state |
| Stateful firewall | Track connections and enforce policy | Connection state does not establish application permission |
| IDS | Detect and report suspicious activity | A passive IDS does not directly block traffic |
| IPS | Inspect traffic inline and potentially block it | Incorrect blocking can disrupt services |
| WAF | Inspect and filter web application traffic | Does not replace application authorization |
| VPN | Provide a logical private connection | Does not automatically grant every internal permission |
| NAT | Translate addresses and possibly ports | Does not replace explicit access policy |

Some firewalls also inspect application-layer traffic.

VPN products may include access policies, but tunnel encryption and resource authorization remain distinct functions.

## 14.2 Example Network Policy

| Source | Destination | Example Decision |
|---|---|---|
| Employee network | Payroll application | Allow required HTTPS traffic |
| Guest network | Internal systems | Deny |
| Internet | Public web service | Allow required public service traffic |
| Application servers | Database | Allow required database connections |
| DMZ | Internal servers | Deny except explicitly approved dependencies |
| VPN users | Internal services | Allow only role-required access |

Rules should have a purpose, owner, and review process.

Use default-deny behavior where appropriate, with explicit required allowances.

# 15. Network Troubleshooting

## 15.1 Common Symptoms

| Symptom | Possible Cause | Next Investigation |
|---|---|---|
| Unexpected 169.254.x.x address | DHCP configuration or communication failure | Check link, DHCP server, relay, pool, and logs |
| Gateway responds but another subnet does not | Routing, filtering, or destination issue | Check routes, firewall policy, and target availability |
| IP-based connection works but hostname fails | DNS or name-resolution issue | Check resolver settings and DNS response |
| TCP connection works but certificate warning appears | TLS validation problem | Check hostname, chain, validity, and clock |
| Gateway MAC changes unexpectedly | Spoofing or legitimate failover | Inspect ARP history and network design |

A symptom suggests where to investigate. It does not establish the root cause by itself.

## 15.2 Layer-by-Layer Method

1. Check physical or wireless connectivity.
2. Check IP configuration.
3. Check the local gateway.
4. Check routing.
5. Check DNS.
6. Check the destination service and port.
7. Check TLS.
8. Check application behavior and authorization.

## 16.3 DNS Example

```bash
dig example.com A
dig example.com AAAA
```

## 16.4 HTTP Example

```bash
curl -I https://example.com
```

This requests response headers using HEAD. The server may handle HEAD differently from GET.

## 16.5 Packet Capture Example

On your own lab system or an authorized environment:

```bash
sudo tcpdump -i any -nn 'tcp port 443'
```

This example uses Linux's `any` capture interface.

Capturing TCP 443 will not ordinarily reveal encrypted HTTP contents without suitable decryption information.

# 17. Networking in SOC Investigations

## 17.1 Useful Network Evidence

- Source and destination IP.
- Source and destination port.
- Transport protocol.
- Connection timing.
- DNS names and responses.
- Bytes transferred.
- TCP flags.
- Firewall decisions.
- TLS metadata where available.

## 17.2 Example Investigation

An endpoint repeatedly connects to an unfamiliar destination on TCP 443.

Do not conclude that the traffic is safe because it uses HTTPS.

Investigate:

- Which process initiated it.
- Which hostname was resolved.
- Whether the destination is expected.
- Timing and transfer patterns.
- Related endpoint alerts.
- Whether the connection was permitted.
- Application metadata available from approved monitoring.

# 18. Practice Questions

## Question 1: Why Are MAC and IP Addresses Both Used?

IP addressing supports communication across networks.

On Ethernet, MAC addressing supports delivery to the next hop on the local link.

## Question 2: How Many Conventional Usable Hosts Are in /24, /26, and /30?

- /24: 254.
- /26: 62.
- /30: 2.

Different rules apply to /31 links, /32 addresses, and some managed cloud networks.

## Question 3: Are 10.1.1.10/24 and 10.1.2.10/24 in the Same Subnet?

No.

Their network addresses are:

```text
10.1.1.0/24
10.1.2.0/24
```

## Question 4: Why Is NAT Not a Substitute for a Firewall?

NAT translates addresses.

A firewall enforces an explicit traffic policy.

## Question 5: What Does a SYN Flood Attempt to Exhaust?

Resources associated with incomplete TCP connection establishment, such as the half-open connection backlog.

## Question 6: Does a Successful Ping Prove a Website Works?

No.

It shows that an ICMP exchange succeeded at that time.

DNS, TCP, TLS, HTTP, and application authorization may still fail.

## Question 7: Does Port 443 Prove Traffic Is Safe?

No.

A port number does not establish the application, legitimacy, or safety of the traffic.

## Question 8: Does TLS Grant Access to Private Data?

No.

Application authorization determines access to private resources.

# 19. Key Takeaways

- Different protocols perform different jobs.
- Layer 2 forwarding uses link-layer information.
- IP routing determines packet paths between networks.
- Link frames change across routed hops.
- Subnet membership depends on addressing and prefix configuration.
- TCP provides reliable ordered transport.
- UDP can support applications that implement their own reliability.
- DNS, routing, transport, TLS, and applications can fail independently.
- NAT, firewalls, VPNs, and application authorization serve different purposes.
- Investigate traffic using evidence rather than port numbers alone.
