# Linux Commands Handbook: Basic to Advanced

## How to Use This Guide in Kaggle

This handbook covers:

- Basic Linux navigation
- Files and directories
- Text processing
- Permissions
- Processes
- Networking
- Package management
- Shell scripting
- System administration
- Security and troubleshooting

**Important:** Commands shown in a Markdown cell are displayed but not executed.

Run a Linux command inside a Kaggle code cell:

```bash
!pwd
```

Run several Linux commands inside one Kaggle code cell:

```bash
%%bash
pwd
ls -la
python3 --version
```

**Warning:** Do not execute destructive commands unless you understand exactly what they do.

## Section 14: File Permissions

Linux permissions use:

- `r` for read
- `w` for write
- `x` for execute

Permission numbers use:

- `4` for read
- `2` for write
- `1` for execute

### Display file permissions

```bash
ls -l
```

### Give the owner execute permission

```bash
chmod u+x script.sh
```

### Make a script executable

```bash
chmod +x script.sh
```

### Allow only the owner to read and write a file

```bash
chmod 600 secret.txt
```

### Set standard permissions for a normal file

```bash
chmod 644 document.txt
```

### Set standard permissions for an executable or directory

```bash
chmod 755 script.sh
```

### Remove write permission from everyone

```bash
chmod a-w document.txt
```

### Change a file’s owner

```bash
sudo chown student document.txt
```

### Change a file’s owner and group

```bash
sudo chown student:developers document.txt
```

### Display your user and group identifiers

```bash
id
```

### Display the current permission mask

```bash
umask
```
